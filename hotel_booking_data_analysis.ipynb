### Variables Description

hotel: Indicates the type of hotel - either a resort hotel (H1) or a city hotel (H2).

arrival_date_week_number: Week number of the arrival date.

arrival_date_day_of_month: Day of the month of the arrival date.

stays_in_weekend_nights: Number of weekend nights (Saturday or Sunday) the guest stayed or booked to stay at the hotel.

stays_in_week_nights: Number of week nights (Monday to Friday) the guest stayed or booked to stay at the hotel.

adults: Number of adults included in the booking.

children: Number of children included in the booking.

babies: Number of babies included in the booking.

meal: Type of meal option opted for.

country: Country code of the guest.

market_segment: Market segment of the booking (e.g., corporate, direct, travel agents).

distribution_channel: Channel through which the booking was made (e.g., corporate, direct, travel agents).

is_repeated_guest: Indicates if the guest is a repeated guest (1) or not (0).

previous_cancellations: Count of previous booking cancellations by the guest.

previous_bookings_not_canceled: Count of previous bookings not canceled by the guest.

reserved_room_type: Type of room reserved by the guest.

assigned_room_type: Type of room assigned to the guest.

booking_changes: Count of changes made to the booking.

deposit_type: Type of deposit made for the booking.

agent: ID of the travel agent through whom the booking was made.

days_in_waiting_list: Number of days the booking was in the waiting list before it was confirmed.

customer_type: Type of customer (e.g., transient, contract, group).

adr: Average daily rate (i.e., average revenue per available room).

required_car_parking_spaces: Number of car parking spaces required by the guest.

total_of_special_requests: Number of additional special requests made by the guest.

reservation_status: Status of the reservation (e.g., canceled, check-out, no-show).

reservation_status_date: Date of the specific reservation status.

## Visualization of conceled booking

```python

# Count the number of canceled and non-canceled bookings
booking_counts = df['is_canceled'].value_counts()

# Plot the graph
plt.figure(figsize=(8, 6))
booking_counts.plot(kind='bar', color=['skyblue', 'lightgreen'])
plt.title('Total Number of Bookings vs Canceled or Non-Canceled Bookings')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks([0, 1], ['Non-Canceled', 'Canceled'], rotation=0)
plt.show()

# Comparison of canceled and non-canceled bookings
canceled_count = booking_counts[1]
non_canceled_count = booking_counts[0]
print(f"Number of canceled bookings: {canceled_count}")
print(f"Number of non-canceled bookings: {non_canceled_count}")

# Statement comparing canceled and non-canceled bookings
if canceled_count > non_canceled_count:
    print("There are more canceled bookings compared to non-canceled bookings.")
elif canceled_count < non_canceled_count:
    print("There are more non-canceled bookings compared to canceled bookings.")
else:
    print("The number of canceled bookings is equal to the number of non-canceled bookings.")
```

```python
import matplotlib.pyplot as plt

# Calculate the percentage of canceled and non-canceled bookings
total_bookings = df.shape[0]
percentage_canceled = (df['is_canceled'].sum() / total_bookings) * 100
percentage_non_canceled = ((total_bookings - df['is_canceled'].sum()) / total_bookings) * 100

# Create a pie chart
labels = ['Canceled', 'Non-Canceled']
sizes = [percentage_canceled, percentage_non_canceled]
colors = ['lightcoral', 'lightskyblue']
explode = (0.1, 0)  # explode the 1st slice (i.e., 'Canceled')

plt.figure(figsize=(8, 6))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True, startangle=140)
plt.title('Percentage of Canceled and Non-Canceled Bookings')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.show()

# Statement comparing percentages of canceled and non-canceled bookings
print(f"Percentage of canceled bookings: {percentage_canceled:.2f}%")
print(f"Percentage of non-canceled bookings: {percentage_non_canceled:.2f}%")

if percentage_canceled > percentage_non_canceled:
    print("The percentage of canceled bookings is higher than non-canceled bookings.")
elif percentage_canceled < percentage_non_canceled:
    print("The percentage of non-canceled bookings is higher than canceled bookings.")
else:
    print("The percentage of canceled bookings is equal to the percentage of non-canceled bookings.")
```

## Relationship between Previous Cancellations and current Cancellations

```python

# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(df['previous_cancellations'], df['is_canceled'], alpha=0.5)
plt.title('Relationship between Previous Cancellations and Current Cancellations')
plt.xlabel('Previous Cancellations')
plt.ylabel('Current Cancellations')
plt.grid(True)
plt.show()

# Calculate the correlation coefficient
correlation = df['previous_cancellations'].corr(df['is_canceled'])

# Statement line with comparison
if correlation > 0:
    print("There is a positive correlation between the number of previous cancellations and current cancellations.")
elif correlation < 0:
    print("There is a negative correlation between the number of previous cancellations and current cancellations.")
else:
    print("There is no correlation between the number of previous cancellations and current cancellations.")
```

```python
from scipy import stats

# Separate the lead time for canceled and non-canceled bookings
lead_time_canceled = df[df['is_canceled'] == 1]['lead_time']
lead_time_not_canceled = df[df['is_canceled'] == 0]['lead_time']

# Perform two-sample t-test
t_statistic, p_value = stats.ttest_ind(lead_time_canceled, lead_time_not_canceled)

# Define significance level (alpha)
alpha = 0.05

# Print the results
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < alpha:
    print("Reject the null hypothesis. There is a significant difference in average lead time between canceled and non-canceled bookings.")
else:
    print("Fail to reject the null hypothesis. There is no significant difference in average lead time between canceled and non-canceled bookings.")

```

The statistical test performed to obtain the P-value is the independent two-sample t-test.

##### Why did you choose the specific statistical test?

The independent two-sample t-test is chosen because it is appropriate for comparing the means of two independent groups to determine if there is a significant difference between them. In this case, we are comparing the average daily rates (ADR) between peak season and off-peak season bookings, which are two independent groups. The t-test helps us assess whether the observed difference in ADR between the two seasons is statistically significant or if it could have occurred by random chance.

### 1. Handling Missing Values

```python
# Handling Missing Values & Missing Value Imputation

# Identify missing values
missing_values = df.isnull().sum()

# Remove rows with missing values (if necessary)
# df.dropna(inplace=True)

# Imputation
# Numeric columns: Replace missing values with mean or median
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns
for column in numeric_columns:
    df[column].fillna(df[column].mean(), inplace=True)

# Categorical columns: Replace missing values with mode
categorical_columns = df.select_dtypes(include=['object']).columns
for column in categorical_columns:
    df[column].fillna(df[column].mode()[0], inplace=True)

# Evaluate imputation quality
imputation_quality = df.isnull().sum()

# Display the imputed DataFrame
print(df.head())

# Display the evaluation of imputation quality
print(imputation_quality)
```

#### What all missing value imputation techniques have you used and why did you use those techniques?

* **Mean imputation for numeric columns:** This technique replaces missing values with the mean of the respective column. It is a simple approach that works well when the data distribution is approximately normal and there are no extreme outliers. Mean imputation helps preserve the overall distribution of the data and can be computationally efficient.
* **Mode imputation for categorical columns:** This technique replaces missing values with the mode (most frequent value) of the respective column. Mode imputation is suitable for categorical variables where there is a clear dominant category. It helps maintain the integrity of categorical distributions and is straightforward to implement.

I chose these techniques because they are simple, widely used, and suitable for handling missing values in both numeric and categorical columns. Additionally, these techniques are often the first step in the missing value imputation process and serve as a baseline approach before exploring more advanced methods. However, it's important to note that the choice of imputation technique may vary depending on the nature of the data, the amount of missingness, and the specific requirements of the analysis.

##### What all outlier treatment techniques have you used and why did you use those techniques?

Interquartile Range (IQR): The IQR is a measure of statistical dispersion that captures the spread of the middle 50% of the data. Outliers are defined as observations that fall below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR, where Q1 and Q3 are the first and third quartiles, respectively. This method is robust to extreme values and works well for symmetric distributions.
Reasons for using the IQR method:

**Robustness**: The IQR method is less sensitive to extreme values compared to methods like z-scores, making it suitable for datasets with skewed distributions or presence of outliers.

**Simplicity**: Calculating the IQR and defining the bounds for outliers is straightforward and easy to implement.

**Preservation of Data:** Instead of imputing or modifying outlier values, this method removes them entirely, preserving the integrity of the remaining data.
